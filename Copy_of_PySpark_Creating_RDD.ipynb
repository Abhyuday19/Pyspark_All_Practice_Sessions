```python
## Making a Function and that will do multiple tranformation and having there results
def transformRDD(customer):
  words = customer.split(" , ")
  # Convert male to 0 and female to 1
  if words[2] == "Male" :
    words[2] = "0"
  else :
    words[2] = "1"
  # Convert N to 0 and Y to 1 for the purchased value
  if words[4] == "N" :
    words[4] = "0"
  else:
    words[4] = "1"
  ##Convert Country to upper case
  words[3] = words[3].upper()
  return ",".join(words)
```

```python
my_csv_transformed_rdd = my_csv_rdd.map(transformRDD)
```

```python
# 1. Point specifically to the downloaded retailstore.csv instead of the entire sample_data folder
my_csv_rdd = sc.textFile("retailstore.csv")

# 2. Update the transformation function to split by standard comma ',' and add a length check to avoid IndexError
def transformRDD_safe(customer):
    words = [w.strip() for w in customer.split(",")]

    # Skip header or empty/invalid rows safely
    if len(words) < 5 or words[0] == "Val1":
        return customer

    # Convert male to 0 and female to 1
    if words[2].lower() == "male":
        words[2] = "0"
    else:
        words[2] = "1"

    # Convert N to 0 and Y to 1 for the purchased value
    if words[4].upper() == "N":
        words[4] = "0"
    else:
        words[4] = "1"

    # Convert Country to upper case
    words[3] = words[3].upper()
    return ",".join(words)

# 3. Apply the safe transformation and collect the output
my_csv_transformed_rdd = my_csv_rdd.map(transformRDD_safe)
my_csv_transformed_rdd.collect()
```
