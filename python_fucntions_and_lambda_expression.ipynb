```python
def calculate_sum_multiplication(a,b):
  return a+b , a*b
```

```python
var_sum , var_multipl = calculate_sum_multiplication(10,3)
```

```python
print(var_sum)
print(var_multipl)
```

```python
## the lmabda expression helps to perform operations in concie (in small manner)
my_var = lambda x: x+100
print(my_var(4))
```

```python
my_var_1 = lambda x,y : x+y
print(my_var_1(4,5))
```

```python
## the lambda fucntion with the multiple operations and multiple expressions and multiple output
## A lambda function can perform multiple independent operations on the same arguments and return multiple results using a tuple.
my_var_2 = lambda a,b : (a+b, a*b, a/b, a-b)
print(my_var_2(15.6,5.9))
```
